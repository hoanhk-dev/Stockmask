```python
import os
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

os.environ["OPENAI_API_KEY"] = "proxypal-local"
os.environ["OPENAI_API_BASE"] = "http://localhost:8317/v1"

llm = ChatOpenAI(
    model="gpt-4", 
    temperature=0.7,
)

# 4. Create a Prompt Template
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an intelligent and creative AI assistant. Reply in Vietnamese."),
    ("user", "{topic}")
])

# 5. Build the LCEL (LangChain Expression Language) Chain
chain = prompt | llm | StrOutputParser()

print(chain)
```

```python
# ChatGPT Version - Using OpenAI API directly
import os
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

os.environ["OPENAI_API_KEY"] = "proxypal-local"
os.environ["OPENAI_API_BASE"] = "http://localhost:8317/v1"

llm_chatgpt = ChatOpenAI(
    model="gpt-4", 
    temperature=0.7,
)

# Create a Prompt Template
prompt_chatgpt = ChatPromptTemplate.from_messages([
    ("system", "You are an intelligent and creative AI assistant. Reply in Vietnamese."),
    ("user", "{topic}")
])

# Build the LCEL Chain with ChatGPT
chain_chatgpt = prompt_chatgpt | llm_chatgpt | StrOutputParser()

print(chain_chatgpt)
```
